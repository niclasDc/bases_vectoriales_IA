```python

#* Para instalar pinecone se usa: pip install pinecone

from dotenv import load_dotenv
import os
from pinecone import Pinecone, ServerlessSpec
load_dotenv()
PINECONE_API_KEY = os.getenv("PINECONE_API_KEY")
PINECONE_HOST = os.getenv("PINECONE_HOST")
```

```python
pinecone_client = Pinecone(PINECONE_API_KEY)
```

```python
index_name = "indice-prueba"

if not pinecone_client.has_index(index_name):
    pinecone_client.create_index_for_model(
        name=index_name,
        cloud="aws",
        region="us-east-1",
        embed={
            "model": "llama-text-embed-v2",
            "field_map": {"text": "chunk_test"}
        }
    )
```

```python

#* Obtenemos el indice usando el host
index = pinecone_client.Index(host=PINECONE_HOST)
```

```python

# * Ahora usamos el indice para insertar datos
index.upsert_records(
    namespace='clientes-ventas',
    records=[
        {
            "_id": "r1",
            "chunk_test": "El cliente menciona que el tiempo de entrega fue más rápido de lo esperado y quedó satisfecho.",
            "categoria": "satisfaccion_cliente"
        },
        {
            "_id": "r2",
            "chunk_test": "La tienda registró un aumento del 12%en ventas de productos premium durante este trimestre.",
            "categoria": "ventas"
        },
        {
            "_id": "r3",
            "chunk_test": "Un cliente expresó que el soporte técnico resolvió su problema en menos de 10 minutos.",
            "categoria": "soporte_tecnico"
        },
        {
            "_id": "r4",
            "chunk_test": "El producto tuvo una alta demanda en la región norte, superando las metas de ventas proyectadas.",
            "categoria": "demanda_regional"
        }
    ]
)

print("Registros insertados correctamente en el índice de Pinecone.")
```
