```python

# Librerías necesarias

import numpy as np
import matplotlib.pyplot as plt
from numpy.linalg import norm
from mpl_toolkits.mplot3d import Axes3D
```

```python
# Definición de vectores de ejemplo

# Se va a realizar una representación de una búsqueda de recomendación
# Dimensiones: [programación, finanzas, creatividad]

A = np.array([0.5, 0.4, 0.1]) # Curso A: "Introducción a Python"

B = np.array([0.5, 0.8, 0.1]) # Curso B: "Python avanzado para finanzas"

C = np.array([0.1, 0.1, 1.0]) # Curso C: "Diseño gráfico con Canva"
```

```python
# Visualizamos los vectores en un espacio 3D

fig = plt.figure(figsize=(8, 7))
ax = fig.add_subplot(111, projection='3d')

ax.quiver(0, 0, 0, A[0], A[1], A[2], color='r', linewidth=2, arrow_length_ratio=0.1, label='Curso A: Introducción a Python')

ax.quiver(0, 0, 0, B[0], B[1], B[2], color='g', linewidth=2, arrow_length_ratio=0.1, label='Curso B: Python avanzado para finanzas')

ax.quiver(0, 0, 0, C[0], C[1], C[2], color='b', linewidth=2, arrow_length_ratio=0.1, label='Curso C: Diseño gráfico con Canva')

ax.set_xlim([0, 1.1])
ax.set_ylim([0, 1.1])
ax.set_zlim([0, 1.1])

ax.set_xlabel('Programación / Python', fontsize=11, labelpad=10)
ax.set_ylabel('Finanzas / Negocio', fontsize=11, labelpad=10)
ax.set_zlabel('Creatividad / Diseño', fontsize=11, labelpad=15)

ax.set_box_aspect([1,1,1.2])
ax.view_init(elev=20, azim=40)

ax.set_title('Representación de Cursos en un Espacio Vectorial 3D', fontsize=14, pad=20)
ax.legend(loc='upper left')  

plt.tight_layout()
plt.show()
```

```python
# Similitud Coseno entre los cursos

cos_AB = np.dot(A, B) / (norm(A) * norm(B))
cos_AC = np.dot(A, C) / (norm(A) * norm(C))
cos_BC = np.dot(B, C) / (norm(B) * norm(C))

print("Resultados de Similitud Coseno:")
print(f"Similitud Coseno entre Curso A y Curso B: {cos_AB:.3f}")
print(f"Similitud Coseno entre Curso A y Curso C: {cos_AC:.3f}")
print(f"Similitud Coseno entre Curso B y Curso C: {cos_BC:.3f}")
```

```python
# Distancia Euclidiana entre los cursos
l2_AB = norm(A - B)
l2_AC = norm(A - C)
l2_BC = norm(B - C)

print("Resultados de Distancia Euclidiana:")
print(f"Distancia Euclidiana entre Curso A y Curso B: {l2_AB:.3f}")
print(f"Distancia Euclidiana entre Curso A y Curso C: {l2_AC:.3f}")
print(f"Distancia Euclidiana entre Curso B y Curso C: {l2_BC:.3f}")
```

```python
# Producto punto entre los cursos
dot_AB = np.dot(A, B)
dot_AC = np.dot(A, C)
dot_BC = np.dot(B, C)

print("Resultados de Producto Punto:")
print(f"Producto Punto entre Curso A y Curso B: {dot_AB:.3f}")
print(f"Producto Punto entre Curso A y Curso C: {dot_AC:.3f}")
print(f"Producto Punto entre Curso B y Curso C: {dot_BC:.3f}")
```
